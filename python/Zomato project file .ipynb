```python
numeric_cols = [
    "Delivery_person_Age",
    "Delivery_person_Ratings",
    "Vehicle_condition",
    "multiple_deliveries",
    "distance_km",
    "Time_taken (min)"
]
outlier_report = []

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_report.append({
        "Column": col,
        "Lower_Bound": lower,
        "Upper_Bound": upper,
        "Outlier_Count": outliers,
        "Outlier_%": round(outliers / len(df) * 100, 2)
    })

outlier_report = pd.DataFrame(outlier_report)

outlier_report
```

```python
df_clean = df.copy()
```

```python
df_clean["Delivery_person_Age"] = (
    df_clean["Delivery_person_Age"]
    .fillna(df_clean["Delivery_person_Age"].median())
)
```

```python
df_clean["Delivery_person_Age"].isnull().sum()
```

```python
df_clean["Delivery_person_Ratings"] = (
    df_clean["Delivery_person_Ratings"]
    .fillna(df_clean["Delivery_person_Ratings"].median())
)
```

```python
df_clean["Delivery_person_Ratings"].isnull().sum()
```

```python
df_clean["Time_Orderd"] = pd.to_datetime(
    df_clean["Time_Orderd"],
    format="%H:%M",
    errors="coerce"
)
```

```python
df_clean.isnull().sum()
```

```python
df_clean["Order_Date"] = pd.to_datetime(
    df_clean["Order_Date"],
    format="%d-%m-%Y",
    errors="coerce"
)
```

```python
df_clean.info()
```

```python
df_clean.isnull().sum()
```

```python
df_clean["Day_Name"] = df_clean["Order_Date"].dt.day_name()
```

```python
df_clean["Month"] = df_clean["Order_Date"].dt.month_name()
```

```python
df_clean["Day"] = df_clean["Order_Date"].dt.day
```

```python
df_clean["Is_Weekend"] = df_clean["Day_Name"].isin(
    ["Saturday", "Sunday"]
)
```

```python
df_clean["Order_Hour"] = df_clean["Time_Orderd"].dt.hour
```

```python
def time_of_day(hour):
    if pd.isna(hour):
        return "Unknown"
    elif 6 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 21:
        return "Evening"
    else:
        return "Night"

df_clean["Time_of_Day"] = df_clean["Order_Hour"].apply(time_of_day)
```

```python
df_clean["Peak_Hour"] = df_clean["Order_Hour"].isin(
    [19, 20, 21]
)
```

```python
df_clean["Delivery_Performance"] = pd.cut(
    df_clean["Time_taken (min)"],
    bins=[0, 20, 30, 40, 60],
    labels=["Fast", "Normal", "Slow", "Very Slow"]
)
```

```python
p95 = df_clean["Time_taken (min)"].quantile(0.95)

df_clean["Extreme_Delivery"] = (
    df_clean["Time_taken (min)"] >= p95
)
```

```python
df_clean.to_csv(
    "zomato_food_delivery_final_cleaned.csv",
    index=False
)
```
